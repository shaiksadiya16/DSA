```python
#write a code to print the sum of digit of a given number
n=int(input("Enter a number: "))
sum=0
while n!=0:
  d=n%10
  sum+=d
  n//=10
print("Sum of digits ",sum)
```

```python
#write a code to print reverse of a given number
n=int(input("Enter a number: "))
rev=0
while n>0:
  d=n%10
  rev=rev*10+d
  n//=10
print("reverse number ",rev)
```

```python
#write a code to print the count of even and odd digits of a given number
num=int(input("enter a number: "))
even=0
odd=0
while num!=0:
  d=num%10
  if d%2==0:
    even+=1
  else:
    odd+=1
  num//=10
print("Even Count",even)
print("odd  Count",odd)
```
