```python
#checking if an element is present in the array or not
#arr=2,5,8,10,15,16 ->16
arr=[]
n=int(input("Enter the size of array: "))
print("Enter elements: ")
for i in range(n):
  arr.append(int(input()))
target=int(input("Enter the element you want to find: "))
if target in arr:
  print(target,"Found")
else:
  print(target,"not found")
```

```python
#index valuee
arr=[]
n=int(input("Enter the size of array: "))
print("Enter elements: ")
for i in range(n):
  arr.append(int(input()))
target=int(input("Enter the element you want to find: "))
if target in arr:
  print(target,"Found at index: ",arr.index(target))
else:
  print(target,"not found")
```

```python
arr=[]
n=int(input("Enter the size of array: "))
print("Enter elements: ")
for i in range(n):
  arr.append(int(input()))
target=int(input("Enter the element you want to find: "))
found=False
for i in range(n):
  if arr[i]==target:
    print(target,"Found at index: ",i)
    found=True
    break
if found==False:
  print(target,"not found")
```

```python
arr=[]
n=int(input("Enter the size of array: "))
print("Enter elements: ")
for i in range(n):
  arr.append(int(input()))
target=int(input("Enter the element you want to find: "))

# Initialize a counter for the target element
count_of_target = 0

for i in range(n):
  if arr[i]==target:
    count_of_target += 1 # Increment the counter

# Print the total count after the loop
if count_of_target > 0:
  print(f"Count of {target} is {count_of_target}")
else:
  print(f"{target} not found")
```

```python
r=int(input("Enter the number of rows: "))
c=int(input("Enter the number of columns: "))
arr=[]
print("Enter elements: ")
for i in range(r):
  row=[] # Initialize a new list for each row6
  for j in range(c):
    row.append(int(input()))
  arr.append(row)
target=int(input("Enter the element u want to search: "))
found =False
for i in range(r):
  for j in range(c):
    if arr[i][j]==target:
      print("Target coordinates are ",i,j)
      found=True
      break
if found == False:
  print("Element not found....")
```
