```python
#Time complexity and space complexity
#constant time o(1)
num = int(input("Enter the number: "))
print("Square of the number ", num*num)
```

```python
#linear time o(n)
n = int(input("Enter the number: "))
for i in range (1,n+1):
  print(i)
```

```python
#Quadratic time o(n^2)
n = int(input("Enter the number: "))
for i in range (1,n+1):
  for j in range(1,n+1):
    print(i,end=' ')
  print()
```

```python
#logarithmic time o(log n)
n = int(input("Enter the number: "))
while n>1:
  print(n)
  n//=2
```

```python
#independent time o(2n)
n=int(input("Enter the number: "))
for i in range(n):
  print(i,end=' ')
for j in range(n):
  print(j,end=' ')
```

```python
#Best time o(n//2)
n=int(input("Enter the number: "))
for i in range(n//2):
  print(i,end=' ')
```

```python
#constant + linear time o(n+2)
n=int(input("Enter the number: "))
for i in range(n):
  print(i,end=' ')
print("Hi")
print("Students")

```

```python
#Exponential time o(2^n)
def abc(n):
  if n==0:
    return
  print(n)
  abc(n-1)
  abc(n-1)
n=int(input("Entera number: "))
abc(n)
```
