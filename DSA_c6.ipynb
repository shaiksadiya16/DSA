```python
#Code to read n values and traverse those values and print them sum
n=int(input("Enter size of array:"))
arr=[]
for i in range(n):
    num=int(input("Enter a value:"))
    arr.append(num)
print("Array elements:")
for i in range(n):
    print(arr[i],end=' ')
print(*arr)
```

```python
#find the sum of array
n=int(input("Enter size of array:"))
arr=[]
sum=0
for i in range(n):
    num=int(input("Enter a value:"))
    arr.append(num)
print("Array elements:")
for i in range(n):
    sum=sum+arr[i]

print(sum)
```

```python
#find the average of array
n=int(input("Enter size of array:"))
arr=[]
sum=0
for i in range(n):
    num=int(input("Enter a value:"))
    arr.append(num)
print("Array elements:")
for i in range(n):
    sum=sum+arr[i]
    avg=sum/n
print(sum)
print(avg)
```

```python
#find the even/odd numbers in prime
n=int(input("Enter size of array:"))
arr=[]
evencount,oddcount=0,0
for i in range(n):
    num=int(input("Enter a value:"))
    arr.append(num)
for i in range(n):
    if arr[i]%2==0:
        evencount+=1
    else:
        oddcount+=1
print("evencount:",evencount)
print("oddcount:",oddcount)
```

```python
#find the positive/negative numbers in prime
n=int(input("Enter size of array:"))
arr=[]
positive,negative=0,0
for i in range(n):
    num=int(input("Enter a value:"))
    arr.append(num)
for i in range(n):
    if arr[i]>0:
        positive+=1
    else:
        negative+=1
print("positive:",positive)
print("negative:",negative)
```
